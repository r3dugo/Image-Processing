## Normalising Inflection Intensity
For each inflection intensity file, normalise against uniform plate intensity.

Since uniform inflection times is roughly uniform, no need to normalise per experiment.

```python
plates = ["plate-1", "plate-2", "plate-3", "plate-4", "plate-5","plate-6"]
```

```python
uniform_raw = pd.read_csv("../experiments/uniform-plate/output/uniform_intensity.csv", index_col=0)

for plate in plates:
    path = os.path.join("..", "experiments", plate, "output")
    exp = pd.read_csv(os.path.join(path, "intensity_points.csv"), index_col=0)

    uniform = uniform_raw.copy()

    if uniform.shape != exp.shape:
        raise ValueError(
            f"Shape mismatch for {plate}: experiment {exp.shape}, uniform {uniform.shape}"
        )

    # Align by physical plate position, then copy experiment labels
    uniform.index = exp.index
    uniform.columns = exp.columns

    max_uniform = uniform.max().max()
    correction_factor = uniform / max_uniform

    corrected = exp / correction_factor.replace(0, np.nan)

    output_dir = os.path.join("output", "intensity")
    os.makedirs(output_dir, exist_ok=True)

    output_path = os.path.join(output_dir, f"{plate}-corrected.csv")
    corrected.to_csv(output_path, index_label="CN")

    print(f"Saved {output_path}")
```
